```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy import integrate
from math import *

def gauss(x, A, x0, sig, z0): #The formula for Gaussian distribution
   return A*np.exp(-(x-x0)**2/(2*sig**2))+z0

Ai = input("A value for A")
x0i = input("A value for x0")
sigi = input("A value for sigma")
z0i = input("A value for z0")

A = float(Ai)
x0 = float(x0i)
sig = float(sigi)
z0 = float(z0i)

x = np.linspace(-10, 10, 200) #To generate 200 samples on the gaussian distriution
plt.plot(x, gauss(x, A, x0, sig, z0)) #To plot the samples

ai=input("lower boundery")
bi=input("upper boundery")

a = float(ai)
b = float(bi)

def Area(x, A, x0, sig, z0): #The Area formula
   return A*np.exp(-(x-x0)**2/(2*sig**2))
t3, err3 = integrate.quad( Area, a, b, args=(A, x0, sig, z0)) #The second integration using the Area formula
print(f"the area calculated with the Area formula between a and b is  is {t3}")

x_fill = x[(x >= a) & (x <= b)] #The bounderies for where to fill
y_fill = gauss(x_fill, A, x0, sig, z0)
plt.fill_between(x_fill, y_fill, color='orange', alpha=0.3,label=f"The area is: {t3}")
plt.legend()

plt.show()

```
