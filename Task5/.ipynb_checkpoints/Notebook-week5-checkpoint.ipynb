Question 5.6 : Array concatenation

```python
#Q1
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.concatenate([a,b])
print(c) #We get a list in 1D

#Q2
import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])


c2 = np.concatenate((a, b)) # Gives a matrix of 2 columns and 4 rows. It enters after every list
print("c2", c2)
c3 = np.concatenate((a, b), axis=0)  # Gives a matrix of 2 columns and 4 rows. So puts the 2 2x2 matrices under eachother
print("c3", c3)
c4 = np.concatenate((a, b), axis=1) # Gives a matrix of 4 columns and 2 rows. So puts the 2 2x2 matrices next to eachother
print("c4", c4)
```

Question 5.9 : Exploring artificial image colors

```python
from imageio import imwrite, imread #To be able to use images
from matplotlib.pyplot import figure, show
import matplotlib.pyplot as plt

m101 = imread("./Pictures/m101BW.jpg") #rename the image tot m101


plt.imshow(m101, cmap='jet') #color is jet
plt.contour(m101, levels=[300,400,500,550,620], colors=['y','g','purple','b','r'])
plt.show()

treshold = np.percentile(m101,90) 
filterdata = np.where(m101>=treshold,m101,0) #Filters out the highest 10 percent

plt.imshow(filterdata, cmap='jet')
plt.show() # Shows the second graph
```

Question 5.12 : Getting acquainted with classes

```python
from scipy.constants import c
class Galaxy:    
    """Class for redshift calculation of galaxies"""  
    c = c # Make the speed of light a class variable
    caH_0 = 3968.5      # Calcium H line in rest frame
    
    def __init__(self, name, calciumHline=caH_0):
        self.name = name
        self.caH = calciumHline
    def redshift(self):        
        z = (self.caH - self.caH_0)/self.caH
        return z
    def velocity(self): #Define the formula for velocity
        v = (self.redshift()*c)/1000 #Devided by 1000 to get the velocity in km/s
        return v

galS = Galaxy("myGalaxy", 2345) #Name my object
print("The velocity is:", galS.velocity(), "km/s") #Calculate the velocity and print it
```

Question 5.13 : Ellipse plotter with astronomical parameters


```python
import numpy as np
from matplotlib.pyplot import figure, show
from matplotlib.patches import Ellipse
import matplotlib.pyplot as plt


Nx = 400
Ny = 300
data = np.random.random((Ny,Nx)) #plot all random generated points


fig = figure()
frame = fig.add_subplot(1,1,1, aspect=1)
xcenter = 200 #Set the ellipse in the center
ycenter = 150
width = 2*50; height= 2*100 #give the given width and height
angle = 30.0 #And angle
e1 = Ellipse((xcenter, ycenter), width, height, angle=angle, fill=False) #Make the ellipse
frame.add_patch(e1) #plot the ellipse
plt.imshow(data, interpolation='none', origin="lower", cmap='gray') #plot the random generated data
```
